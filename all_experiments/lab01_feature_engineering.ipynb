## Part B — Feature engineering
Derive family size, isolation, title, and log fare from prediction-time information. Each feature must exist before the outcome.

```python
df["FamilySize"] = df["SibSp"] + df["Parch"] + 1
df["IsAlone"] = (df["FamilySize"] == 1).astype(int)
df["Title"] = df["Name"].str.extract(r",\s*([^\.]+)\.")[0].str.strip()
df["Title"] = df["Title"].where(df["Title"].isin(["Mr", "Mrs", "Miss", "Master"]), "Other")
df["LogFare"] = np.log1p(df["Fare"])

print(df[["Name", "Title", "FamilySize", "IsAlone", "Fare", "LogFare"]].head())
print("\nTitle counts:\n", df["Title"].value_counts())
```

Compare survival rates across engineered features.

```python
fig, axes = plt.subplots(1, 2, figsize=(9.5, 3.4))
df.groupby("Title")["Survived"].mean().sort_values().plot.barh(ax=axes[0])
axes[0].set_xlabel("Survival rate")
axes[0].set_title("Title carries the sex-and-age story in one column")
df.groupby("FamilySize")["Survived"].mean().plot.bar(ax=axes[1], rot=0)
axes[1].set_ylabel("Survival rate")
axes[1].set_title("Small families did best; large ones did worst")
plt.tight_layout()
plt.show()
```

**Inspect.** Survival varies by title and family size; small families fare better than lone passengers or large families.

Compare original and engineered features using the same model and split. Imputation, scaling, and encoding are fitted inside the training pipeline.

```python
def make_preprocess(num_cols, cat_cols):
    return ColumnTransformer([
        ("num", Pipeline([
            ("impute", SimpleImputer(strategy="median")),
            ("scale", StandardScaler()),
        ]), num_cols),
        ("cat", Pipeline([
            ("impute", SimpleImputer(strategy="most_frequent")),
            ("onehot", OneHotEncoder(handle_unknown="ignore")),
        ]), cat_cols),
    ])

y = df["Survived"]
feature_sets = {
    "raw columns": (["Age", "SibSp", "Parch", "Fare"], ["Pclass", "Sex", "Embarked"]),
    "raw + engineered": (["Age", "SibSp", "Parch", "LogFare", "FamilySize", "IsAlone"],
                         ["Pclass", "Sex", "Embarked", "Title"]),
}

for name, (num_cols, cat_cols) in feature_sets.items():
    X = df[num_cols + cat_cols]
    X_train, X_valid, y_train, y_valid = train_test_split(
        X, y, test_size=0.25, random_state=42, stratify=y
    )
    model = Pipeline([
        ("preprocess", make_preprocess(num_cols, cat_cols)),
        ("model", LogisticRegression(max_iter=2000)),
    ]).fit(X_train, y_train)
    auc = roc_auc_score(y_valid, model.predict_proba(X_valid)[:, 1])
    print(f"{name:18s} validation ROC AUC = {auc:.3f}")
```

**Inspect.** Higher validation AUC measures the gain from features, not a different estimator.

## Part D — Feature selection without leakage
Predict G3 for 649 students at enrolment. Exclude later grades G1 and G2, leaving 30 candidate columns.

```python
students = pd.read_csv(DATA / "student_performance_dataset.csv")
print("shape:", students.shape, " missing:", students.isna().sum().sum(),
      " duplicates:", students.duplicated().sum())

target = "G3"
leaky = ["G1", "G2"]
X_s = students.drop(columns=[target] + leaky)
y_s = students[target]
s_num = X_s.select_dtypes(include="number").columns.tolist()
s_cat = X_s.select_dtypes(exclude="number").columns.tolist()
print(len(s_num), "numeric and", len(s_cat), "text features available at enrolment")

Xs_train, Xs_valid, ys_train, ys_valid = train_test_split(X_s, y_s, test_size=0.25, random_state=42)
```

Compare ridge using all features or ten SelectKBest features, and lasso's sparse fit. Selection stays inside training pipelines. A fourth run deliberately adds G2 to expose leakage.

```python
def evaluate(pipeline, X_tr, X_va, label):
    pipeline.fit(X_tr, ys_train)
    mae = mean_absolute_error(ys_valid, pipeline.predict(X_va))
    print(f"{label:32s} validation MAE = {mae:.3f} grade points")
    return mae

results = {}
results["all 30 features"] = evaluate(Pipeline([
    ("preprocess", make_preprocess(s_num, s_cat)),
    ("model", Ridge(alpha=1.0)),
]), Xs_train, Xs_valid, "all 30 features")

results["SelectKBest, k=10"] = evaluate(Pipeline([
    ("preprocess", make_preprocess(s_num, s_cat)),
    ("select", SelectKBest(f_regression, k=10)),
    ("model", Ridge(alpha=1.0)),
]), Xs_train, Xs_valid, "SelectKBest, k=10")

lasso = Pipeline([
    ("preprocess", make_preprocess(s_num, s_cat)),
    ("model", Lasso(alpha=0.1)),
])
results["lasso (selects itself)"] = evaluate(lasso, Xs_train, Xs_valid, "lasso (selects itself)")

leak_num = s_num + ["G2"]
results["all 30 + G2 (leakage)"] = evaluate(Pipeline([
    ("preprocess", make_preprocess(leak_num, s_cat)),
    ("model", Ridge(alpha=1.0)),
]), Xs_train.assign(G2=students.loc[Xs_train.index, "G2"]),
    Xs_valid.assign(G2=students.loc[Xs_valid.index, "G2"]), "all 30 + G2 (leakage)")
```

```python
coefs = pd.Series(lasso.named_steps["model"].coef_,
                  index=lasso.named_steps["preprocess"].get_feature_names_out())
kept = coefs[coefs != 0].sort_values()
print(f"lasso kept {len(kept)} of {len(coefs)} encoded columns")

fig, axes = plt.subplots(1, 2, figsize=(11, 3.8))
axes[0].barh(list(results), list(results.values()))
axes[0].set_xlabel("Validation MAE (grade points)")
axes[0].set_title("Fewer features, same error; a leaked feature, half the error")
axes[1].barh(kept.index, kept.values)
axes[1].axvline(0, color="grey", linewidth=0.8)
axes[1].set_title("What the lasso kept")
plt.tight_layout()
plt.show()
```

**Inspect.** Selected features give similar or better error with fewer columns. Adding G2 cuts error sharply but invalidates an enrolment-time forecast. Lasso's retained school, failure, education, alcohol, and study-time features are associations, not causes.

### 1. Cabin initial and fare per family member (Part B)
Two more prediction-time features:
- **CabinInitial** – first letter of `Cabin` (the deck); missing cabins become `"U"` (unknown).
- **FarePerPerson** – `Fare / FamilySize`, since a family often shared one ticket price.

Same model, same split, same `random_state` as Part B, so any AUC change comes from the features only.

```python
df["CabinInitial"] = df["Cabin"].str[0].fillna("U")
df["FarePerPerson"] = df["Fare"] / df["FamilySize"]
print(df["CabinInitial"].value_counts())

feature_sets_try = {
    "raw columns": feature_sets["raw columns"],
    "raw + engineered": feature_sets["raw + engineered"],
    "+ CabinInitial": (feature_sets["raw + engineered"][0],
                       feature_sets["raw + engineered"][1] + ["CabinInitial"]),
    "+ FarePerPerson": (feature_sets["raw + engineered"][0] + ["FarePerPerson"],
                        feature_sets["raw + engineered"][1]),
    "+ both": (feature_sets["raw + engineered"][0] + ["FarePerPerson"],
               feature_sets["raw + engineered"][1] + ["CabinInitial"]),
}

try_auc = {}
for name, (num_cols, cat_cols) in feature_sets_try.items():
    X = df[num_cols + cat_cols]
    X_train, X_valid, y_train, y_valid = train_test_split(
        X, y, test_size=0.25, random_state=42, stratify=y
    )
    model = Pipeline([
        ("preprocess", make_preprocess(num_cols, cat_cols)),
        ("model", LogisticRegression(max_iter=2000)),
    ]).fit(X_train, y_train)
    try_auc[name] = roc_auc_score(y_valid, model.predict_proba(X_valid)[:, 1])
    print(f"{name:18s} validation ROC AUC = {try_auc[name]:.3f}")
```

**Result.** Neither new feature helped on this split. `FarePerPerson` left AUC unchanged (0.872), probably because `LogFare` and `FamilySize` already carry the same information between them. `CabinInitial` lowered AUC slightly (0.872 → 0.868): 687 of 891 cabins are missing (), and rare decks such as G and T have only a handful of passengers, so the extra one-hot columns add noise more than signal. Differences of 0.004 on one 223-row validation split are too small to trust; cross-validation (Lab 2 Part D) would be the proper test. Conclusion: keep the Part B engineered set.

The leaked G2 run (MAE 0.762) remains far lower than any honest setting, a reminder that a big jump in accuracy is often a sign of leakage, not a better model.

## Check your understanding

**1. Why must train/test splitting happen before any preprocessing step that learns from the data?**

Steps like imputation (median), scaling (mean and standard deviation), encoding (which categories exist) and feature selection all *learn numbers from the data*. If they are fitted before the split, those numbers include the validation/test rows, so information about the "unseen" data leaks into training. The model then looks better on the test set than it will on genuinely new data. Splitting first, and putting every learning step inside a `Pipeline` fitted only on the training rows, keeps the test set truly unseen.

**2. What is the difference between an identifier, a feature, and a label?**

- An **identifier** only names a row (e.g. `PassengerId`, tumour `id`). It carries no general information, so a model that uses it just memorises rows; it should be dropped.
- A **feature** is an input the model uses to predict, known at prediction time (e.g. `Age`, `Pclass`, `Sex`).
- A **label** (target) is the answer the model is trained to predict (e.g. `Survived`, `G3`).

**3. Why do you need to know the moment of prediction before you can say whether a feature is leakage?**

Leakage means using information that would not be available when the prediction is actually made. Whether a column is available depends on *when* we predict. In Part D we predict G3 **at enrolment**, so G1 and G2 (grades from later in the year) do not exist yet and using them is leakage, which is why adding G2 made the error look much smaller. If instead we predicted G3 at the end of the second period, G2 would be a legitimate feature. The same column can be valid or leaky depending on the prediction moment.
